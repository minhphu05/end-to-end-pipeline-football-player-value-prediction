# 📊 Phân tích Ma Trận Tương Quan (Correlation Matrix)

## 🎯 1. Giữa 2 Feature Bất Kỳ (Không phải label)

| Mức tương quan | Ý nghĩa                         | Hành động gợi ý                     |
|----------------|----------------------------------|--------------------------------------|
| **r > 0.85**    | Đa cộng tuyến rất mạnh          | → Chọn 1 trong 2 hoặc gộp / PCA      |
| **0.6 < r ≤ 0.85** | Tương quan cao, có thể trùng thông tin | → Cân nhắc giữ 1 cái nếu cần đơn giản |
| **0.3 < r ≤ 0.6** | Trung bình                    | → Thường giữ lại cả 2               |
| **r ≤ 0.3 hoặc âm** | Thấp hoặc ngược chiều        | → Giữ lại, vì bổ sung thông tin khác nhau |

📌 **Lưu ý:** Đặc biệt quan trọng với mô hình tuyến tính như `LinearRegression`, `LogisticRegression`.

---

## 🎯 2. Giữa 1 Feature và Label (Target)

| Mức tương quan | Ý nghĩa                         | Hành động gợi ý                     |
|----------------|----------------------------------|-------------------------------------|
| **r > 0.6**     | Tương quan mạnh với label       | ✅ Nên giữ                          |
| **0.3 < r ≤ 0.6** | Trung bình                    | ⚠️ Xem xét giữ, tùy vào mô hình     |
| **r ≤ 0.3**     | Yếu hoặc không tương quan       | ❌ Có thể loại (nếu không có insight) |
| **r âm mạnh**   | Tương quan ngược chiều mạnh     | ✅ Vẫn giữ nếu hợp lý và có ý nghĩa |

📌 **Mục tiêu:** Giữ lại các feature giúp **mô hình dự đoán tốt hơn** trong **supervised learning**.

---

## ✅ Tóm Gọn Logic

- **Giữa 2 features**:
  - r > 0.85 → đa cộng tuyến → loại/gộp
  - r ~ 0.6–0.85 → cân nhắc giữ 1
  - r thấp hoặc âm → giữ hết

- **Giữa 1 feature và label**:
  - r cao → giữ
  - r thấp → loại nếu không hữu ích
  - r âm → giữ nếu có ý nghĩa (vì âm cũng là thông tin)

---

## 📌 Ghi chú thêm

- **Tree-based models** (RandomForest, XGBoost): ít bị ảnh hưởng bởi đa cộng tuyến.
- **Linear models** (Linear, Logistic): rất nhạy cảm với đa cộng tuyến → nên xử lý.
- Dùng `df.corr()` và `sns.heatmap` để trực quan hóa ma trận tương quan nhanh chóng.
